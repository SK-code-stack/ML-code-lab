# Normalization

## L2 Normalization
Rescales each sample (row) to have unit norm. This type of normalization often used when dealing with text data.It is the squre root of the sum of vactors values.

```python
from sklearn.preprocessing import Normalizer
data = [[0, 1, 2], [2, 0, 0], [0, 0, 1]]
normalizer = Normalizer(norm='l2')
print(normalizer.fit_transform(data))
```

## L1 Normalization
It also rescale each sample but with a different approach, It ensure that the sum of the absolute value is 1 in each row. The L1 norm is calculated as a sum of the absolute vactor.


```python
from sklearn.preprocessing import Normalizer
data = [[1, 1, 1], [1, 1, 0], [1, 0, 0]]
normlizer = Normalizer(norm='l1')
print(normlizer.fit_transform(data))
```

## Log Transformation
Skewed means ka data ak bohat bara or koi bohat chota.

```python
import pandas as pd
import numpy as np

# Example data set with skewed values
df = pd.DataFrame({"values":[1, 5, 10, 50, 100, 2, 6, 12, 60, 120, 3, 7, 15, 70, 150, 422, 8333, 20222, 83330, 200000]})
df
```

```python
import seaborn as sns
sns.histplot(df['values'], kde=True)
```

```python
df['log_values'] = np.log(df['values'])
df.head()
```

```python
sns.histplot(df['log_values'], kde=True)
```
